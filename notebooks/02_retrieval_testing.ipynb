```python
embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5"
)

vector_db = FAISS.load_local(
    "/content/drive/MyDrive/LegalResearch/vectorstore/laws_faiss",
    embeddings,
    allow_dangerous_deserialization=True
)

print("Vector database loaded")
```

```python
retriever = vector_db.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k":5,
        "fetch_k":20,
        "lambda_mult":0.7
    }
)

results = retriever.invoke("What is Article 21?")
```

```python
for i, doc in enumerate(results, 1):

    print("="*80)

    print(f"Result {i}")

    print("Law:", doc.metadata.get("law_name"))

    print("Page:", doc.metadata.get("page")+1)

    print("Source:", doc.metadata.get("source_file"))

    print()

    print(doc.page_content[:1000])
```
